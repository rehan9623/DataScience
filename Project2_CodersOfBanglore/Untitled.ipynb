```python
with open("initialdata.txt", encoding='utf-8') as f:
    data = f.read()
    
```

```python
chunks = data.split("\n\n") 
chunks = [c for c in chunks if len(c)>3]
chunks
```

```python
chunks[5]
```

```python
def parse_chunk(chunk): 
        chunk = chunk.strip()
        lines = chunk.split("\n")
        name = lines[0].replace("Name: ", "")
        posts = lines[1].replace("Posts: ", "")
        followers = lines[2].replace("Followers: ", "")
        bio = lines[3].replace("Bio: ", "")
        
        return {"name": name, "posts": posts, "followers": followers, "bio": bio}


#2nd Way
'''def parse_chunk(chunk): 
        chunk = chunk.strip()
        sep_chunk = chunk.split('\n')
        username = sep_chunk[0]
        no_of_posts = sep_chunk[1]
        no_of_followers = sep_chunk[2]
        no_of_following = sep_chunk[3]
        name = sep_chunk[4]
        if(len(sep_chunk)> 5):
            type_of_page = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_page = "Unknown"
            bio = ""
        # print(username, no_of_posts, no_of_followers, no_of_following, name, type_of_page, bio, sep="\n")
        return {"Username": username, "No. of Post": no_of_posts, "No. of Followers": no_of_followers, 
                "No. of Following": no_of_following, "Name": name, "Type of Page": type_of_page, "BIO": bio}'''
```
