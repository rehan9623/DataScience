## 5.Filtering with Conditions

```python
df[df["IMDb"]>6]
```

```python
df[(df["IMDb"]>7) & (df["Year"]>2020)]
```

```python
df[df["BoxOffice(INR Crore)"]>500]
```

```python
df[(df["BoxOffice(INR Crore)"]>500) | (df["Year"]>2020) ]
```

## - Querying with .query()

```python
import pandas as pd
```

```python
df = pd.read_csv("Bollywood.csv")
```

```python
df.query("Year > 2020")
```

```python
df.query("Year > 2020 and IMDb > 6")
```

```python
col = "Year"
df.query(f"{col} > 2018 ")
```

```python
df.query("`BoxOffice(INR Crore)` > 500 and Genre == 'Action'")
```

```python
df.query("Year > 2019 and Genre == 'Action' and IMDb > 5")
```
